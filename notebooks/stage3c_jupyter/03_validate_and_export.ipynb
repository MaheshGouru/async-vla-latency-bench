# Stage 3C.03 — fail-closed validation and export
A Stage 3D handoff is written only if every determinism and distinctness gate passes.

```python
import hashlib,json,subprocess
from pathlib import Path
R=Path.home()/"async-vla-latency-bench"; PY=Path.home()/"venv-stage1-id/bin/python"; S1=Path.home()/"stage1"; OUT=Path.home()/"stage3c"
cmd=[str(PY),"-m","async_vla_benchmark.scripts.validate_stage3c","--audit",str(OUT/"stage3c_initialization_audit.csv"),"--frozen-variants",str(S1/"stage1_resolved_variants.csv"),"--spec",str(R/"docs/STAGE_3C_INITIALIZATION_AUDIT.md"),"--output-dir",str(OUT)]
validation=subprocess.run(cmd,cwd=R)
report=json.loads((OUT/"stage3c_initialization_audit.json").read_text())
archive=Path.home()/"stage3c_results.tar.gz"; subprocess.run(["tar","-czf",str(archive),"-C",str(Path.home()),"stage3c"],check=True)
print(*sorted(p.name for p in OUT.iterdir()),sep="\n"); print(archive,archive.stat().st_size,"bytes"); print("sha256",hashlib.sha256(archive.read_bytes()).hexdigest()); print("DOWNLOAD THIS ARCHIVE NOW. Stage 3D remains blocked unless status=pass.")
if validation.returncode or report["status"]!="pass": raise SystemExit("STOP: Stage 3C failed closed; preserve the archive and do not run Stage 3D")
assert report["reset_operations"]==144 and report["validated_initializations"]==48
print("PASS: Stage 3C certified 48 deterministic, distinct initializations for Stage 3D handoff")
```
