# Stage 5.01 — setup and pinned policy preflight

Stage 5 uses the same pinned OpenVLA-OFT checkpoint as Stage 4. The 5A0 capability audit can be run statically with the current interpreter; a dedicated OpenVLA-OFT venv is only needed if the gate unexpectedly opens and you want to run `--runtime` model rollouts.

```python
import json, os, subprocess, sys
from pathlib import Path

R = Path.home() / "async-vla-latency-bench"
OFT = Path.home() / "openvla-oft"
P = Path.home() / "LIBERO-plus"
OUT = Path.home() / "stage5"
ANALYSIS = OUT / "analysis"
OUT.mkdir(exist_ok=True)
ANALYSIS.mkdir(exist_ok=True)

# Static audit needs only async_vla_benchmark.
PY = Path(sys.executable).resolve()
# If you later want a runtime OpenVLA-OFT rollout, install the same venv used for Stage 4.
PY_RUN = Path.home() / "venv-stage5-openvla/bin/python"

GIT_SHA = subprocess.run(["git", "-C", str(R), "rev-parse", "HEAD"], capture_output=True, text=True, check=True).stdout.strip()
LIBERO_PLUS_SHA = subprocess.run(["git", "-C", str(P), "rev-parse", "HEAD"], capture_output=True, text=True, check=True).stdout.strip() if (P / ".git").exists() else "unknown"

print("repo_sha", GIT_SHA[:8])
print("libero_plus_sha", LIBERO_PLUS_SHA[:8])
print("static python", PY)
print("runtime python", PY_RUN)
```

```python
# Persist the provenance tokens the Stage 5 scripts need.
prov = {"git_sha": GIT_SHA, "libero_plus_git_sha": LIBERO_PLUS_SHA}
(OUT / "stage5_provenance.json").write_text(json.dumps(prov, indent=2) + "\n")
print("PASS: Stage 5 preflight provenance written to", OUT / "stage5_provenance.json")
```
